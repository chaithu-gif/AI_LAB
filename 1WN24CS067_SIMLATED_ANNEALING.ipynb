```python

import random
import math

def cost(state):
    attacks = 0

    for i in range(8):
        for j in range(i + 1, 8):
            if abs(state[i] - state[j]) == abs(i - j):
                attacks += 1

    return attacks


def get_neighbor(state):
    new_state = state.copy()

    col = random.randint(0, 7)
    row = random.randint(0, 7)

    while row == new_state[col]:
        row = random.randint(0, 7)

    new_state[col] = row

    return new_state


def simulated_annealing():
    state = list(map(int, input("Enter 8 queen positions (0-7): ").split()))

    if len(state) != 8 or any(x < 0 or x > 7 for x in state):
        print("Invalid input! Enter 8 numbers between 0 and 7.")
        return

    temperature = 100
    cooling_rate = 0.99
    minimum_temperature = 0.01

    current_cost = cost(state)

    print("\nInitial State:", state)
    print("Initial Cost:", current_cost)

    while temperature > minimum_temperature:

        if current_cost == 0:
            print("\nGoal State Reached!")
            print("Final State:", state)
            return

        neighbor = get_neighbor(state)
        neighbor_cost = cost(neighbor)

        difference = neighbor_cost - current_cost

        if difference < 0:
            state = neighbor
            current_cost = neighbor_cost

        else:
            probability = math.exp(-difference / temperature)

            if random.random() < probability:
                state = neighbor
                current_cost = neighbor_cost

        print("State:", state, "Cost:", current_cost,
              "Temperature:", round(temperature, 2))

        temperature *= cooling_rate

    if current_cost == 0:
        print("\nGoal State Reached!")
    else:
        print("\nAlgorithm stopped. Solution not found.")

    print("Final State:", state)
    print("Final Cost:", current_cost)


simulated_annealing()
```
