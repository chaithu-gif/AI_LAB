```python
def cost(state):
    attacks = 0

    for i in range(4):
        for j in range(i + 1, 4):
            if abs(state[i] - state[j]) == abs(i - j):
                attacks += 1

    return attacks


def get_neighbors(state):
    neighbors = []

    for i in range(4):
        for j in range(i + 1, 4):
            new_state = state.copy()
            new_state[i], new_state[j] = new_state[j], new_state[i]

            neighbors.append((cost(new_state), i, j, new_state))

    return sorted(neighbors)


def hill_climbing():
    # Replace commas with spaces so it works with both '3,1,2,0' and '3 1 2 0'
    raw_input = input("Enter 4 queen positions (separated by spaces or commas): ")
    state = list(map(int, raw_input.replace(',', ' ').split()))

    if len(state) != 4 or sorted(state) != [0, 1, 2, 3]:
        print("Invalid input! Enter 0, 1, 2, 3 in any order.")
        return

    while True:
        current_cost = cost(state)

        print("\nCurrent State:", state)
        print("Current Cost:", current_cost)

        if current_cost == 0:
            print("Goal State Reached!")
            return

        neighbors = get_neighbors(state)

        best_cost, i, j, best_state = neighbors[0]

        print("\nNeighbor States:")

        for c, x, y, s in neighbors:
            print(s, "Cost:", c)

        if best_cost >= current_cost:
            print("\nAlgorithm Terminated - No Better Neighbor")
            return

        print("\nSwapping:", (i, j))

        state = best_state


hill_climbing()
```
